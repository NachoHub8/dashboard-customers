Cleaning dataset BD Clientes

Resumen de pasos de limpieza
1. Eliminar columnas innecesarias.

2. Renombrar las columnas.

3. Manejar los valores nulos (eliminarlos o rellenarlos).

4. Convertir tipos de datos (fechas, números).

5. Filtrar datos fuera de rango (por ejemplo, edades negativas).

6. Eliminar duplicados.

7. Crear KPIs o métricas útiles para análisis posteriores.

```python
import pandas as pd
```

```python
archivo = (r"C:\Users\Usuario\Desktop\escritorio\DA-DS\datasets\BD-CLIENTES-powerbi_eltiotech.xlsx")
```

```python
df = pd.read_excel(archivo)
```

```python
print(df.head())
```

```python
df = df.dropna() 
```

```python
df['Grupo de clientes'].fillna('Desconocido', inplace=True)
df['Edad'].fillna(0, inplace=True)
```

```python
#comprobando existencia de datos nulos
print(df.isnull().sum())
```

```python
#Convirtiendo el formato de celdas-datos
df['Fecha de alta'] = pd.to_datetime(df['Fecha de alta'], errors='coerce')
```

```python
print(df.dtypes)
```

```python
#Convirtiendo el formato de celdas-datos
df['Edad'] = pd.to_numeric(df['Edad'], errors='coerce')
```

```python
#Filtrando datos fuera de rango
df = df[df['Edad'] >= 0]
```

```python
#Eliminando duplicados
df = df.drop_duplicates()
```

```python
#Crear nuevas columnas de KPIs o métricas
df['edad_promedio_ciudad'] = df.groupby('Ciudad')['Edad'].transform('mean')
```

```python
df['grupo_edad'] = pd.cut(df['Edad'], bins=[0, 18, 35, 50, 100], labels=["0-18", "19-35", "36-50", "51+"], right=False)
```

```python
df.to_excel(r"C:\Users\Usuario\Desktop\escritorio\DA-DS\datasets\moviles_limpio.xlsx", index=False)  # O usamos .to_csv y .csv si queremos tal formato
```
